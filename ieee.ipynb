### Q1. Descriptive analytics: trend from data
**Unit 1** | **Marks:** 5 | **Level:** L3 - Apply  
**Dataset file(s):** `q01_monthly_sales.csv`

Monthly sales (in thousands) are: [120, 135, 128, 150, 165, 158].  
Write a Python program to:  
(a) calculate the average sales,  
(b) print the month-to-month growth % for each month,  
(c) print 'Increasing' or 'Decreasing' for the overall trend by comparing the first and last month.

```python
import csv

with open("q01_monthly_sales.csv") as f:
    rows = list(csv.DictReader(f))
months = [r["month"] for r in rows]
sales = [float(r["sales"]) for r in rows]

print("Average sales:", round(sum(sales) / len(sales), 2))
for i in range(1, len(sales)):
    growth = (sales[i] - sales[i - 1]) / sales[i - 1] * 100
    print(f"{months[i]}: {growth:.2f}%")
print("Overall trend:", "Increasing" if sales[-1] > sales[0] else "Decreasing")
```

### Q2. Objective function and constraints
**Unit 1** | **Marks:** 5 | **Level:** L3 - Apply  
**Dataset file(s):** `q02_products.csv, q02_plans.csv`

A factory makes products A, B and C. Profit per unit: A = 40, B = 30, C = 50. Hours needed per unit: A = 2, B = 1, C = 3.  
Constraints: total hours must not exceed 100, and each product must not exceed 20 units.  
Write a function that takes a plan (units of A, B, C), checks both constraints and returns the total profit.  
Test it with Plan 1 = (20, 10, 10) and Plan 2 = (20, 20, 15) and print whether each plan is feasible and its profit.

```python
import csv

with open("q02_products.csv") as f:
    products = {r["product"]: r for r in csv.DictReader(f)}

MAX_HOURS = 100
MAX_UNITS = 20

def check_plan(plan):
    units = dict(zip(["A", "B", "C"], plan))
    hours = sum(units[p] * int(products[p]["hours_per_unit"]) for p in units)
    profit = sum(units[p] * int(products[p]["profit_per_unit"]) for p in units)
    feasible = hours <= MAX_HOURS and all(u <= MAX_UNITS for u in units.values())
    return feasible, hours, profit

with open("q02_plans.csv") as f:
    for r in csv.DictReader(f):
        plan = (int(r["units_A"]), int(r["units_B"]), int(r["units_C"]))
        feasible, hours, profit = check_plan(plan)
        status = "Feasible" if feasible else "Not feasible"
        print(r["plan"], plan, "Hours:", hours, "Profit:", profit, "->", status)
```

**Unit 1** | **Marks:** 5 | **Level:** L3 - Apply

### Q4. Bias detection
**Unit 1** | **Marks:** 5 | **Level:** L4 - Analyze  
**Dataset file(s):** `q04_loan_groups.csv`

Write a program to calculate the approval rate of each group and the difference between them.  
Print 'Possible bias' if the difference is more than 10 percentage points, otherwise print 'No bias detected'.

```python
import csv

rates = {}
with open("q04_loan_groups.csv") as f:
    for r in csv.DictReader(f):
        rates[r["group"]] = int(r["approved"]) / int(r["applicants"]) * 100

for group, rate in rates.items():
    print(f"{group} approval rate: {rate:.1f}%")
diff = abs(rates["Group A"] - rates["Group B"])
print(f"Difference: {diff:.1f} percentage points")
print("Possible bias" if diff > 10 else "No bias detected")
```

### Q5. Natural language to SQL (rule-based)
**Unit 2** | **Marks:** 5 | **Level:** L3 - Apply  
**Dataset file(s):** `q05_sales.csv`

Table: sales(region, product, amount).  
Write a function nl_to_sql(question) that uses keyword matching (no LLM) and returns an SQL query for:  
(i) 'show all sales' -> SELECT * FROM sales  
(ii) 'total sales by region' -> SELECT region, SUM(amount) FROM sales GROUP BY region  
(iii) 'sales in East' -> SELECT * FROM sales WHERE region = 'East'  
Test the function with one question of each type.

```python
def nl_to_sql(question):
    q = question.lower()
    if "total" in q and "region" in q:
        return "SELECT region, SUM(amount) FROM sales GROUP BY region"
    if " in " in q:
        region = question.split(" in ")[-1].strip().title()
        return f"SELECT * FROM sales WHERE region = '{region}'"
    if "all" in q:
        return "SELECT * FROM sales"
    return "Sorry, question not understood"

for question in ["show all sales", "total sales by region", "sales in East"]:
    print(question, "->", nl_to_sql(question))
```

### Q8. Hallucination check
**Unit 2** | **Marks:** 5 | **Level:** L4 - Analyze  
**Dataset file(s):** `q08_sales.csv`

An AI tool claims: 'Sales grew by 25% from Q1 to Q2.' Actual data: Q1 = 200, Q2 = 220.  
Write a function that calculates the actual growth % and checks the claim (allow a difference of 1 percentage point).  
Print 'Supported' or 'Not supported (actual growth = x%)'.

```python
import csv

with open("q08_sales.csv") as f:
    sales = {r["quarter"]: float(r["sales"]) for r in csv.DictReader(f)}

def check_claim(claimed, q1, q2):
    actual = (q2 - q1) / q1 * 100
    if abs(actual - claimed) <= 1:
        return "Supported"
    return f"Not supported (actual growth = {actual:.1f}%)"

print(check_claim(25, sales["Q1"], sales["Q2"]))
```

### Q11. Similarity search
**Unit 3** | **Marks:** 5 | **Level:** L3 - Apply  
**Dataset file(s):** `q11_vectors.csv`

Write a function to calculate the cosine similarity of two vectors (use only the math module).  
For A = [1, 2, 3], B = [2, 4, 6] and C = [3, 0, 1], print similarity(A, B) and similarity(A, C), and print which of B and C is more similar to A.

```python
import csv
import math

def cosine(a, b):
    dot = sum(x * y for x, y in zip(a, b))
    return dot / (math.sqrt(sum(x * x for x in a)) * math.sqrt(sum(y * y for y in b)))

with open("q11_vectors.csv") as f:
    v = {r["name"]: [float(r["x1"]), float(r["x2"]), float(r["x3"])] for r in csv.DictReader(f)}

sim_ab = cosine(v["A"], v["B"])
sim_ac = cosine(v["A"], v["C"])
print("similarity(A, B) =", round(sim_ab, 4))
print("similarity(A, C) =", round(sim_ac, 4))
print("More similar to A:", "B" if sim_ab > sim_ac else "C")
```

### Q12. Simple retrieval with source
**Unit 3** | **Marks:** 5 | **Level:** L4 - Analyze  
**Dataset file(s):** `q12_documents.csv`

Documents: D1 = 'refund is processed in 5 working days', D2 = 'delivery takes 3 to 5 days', D3 = 'damaged items are replaced free'.  
For the query 'how many days for refund', write a program that counts the common words between the query and each document, and prints the best matching document with its ID as the source.

```python
import csv

query = "how many days for refund"
query_words = set(query.split())

best_id, best_text, best_score = None, "", -1
with open("q12_documents.csv") as f:
    for r in csv.DictReader(f):
        score = len(query_words & set(r["text"].split()))
        print(r["doc_id"], "common words:", score)
        if score > best_score:
            best_id, best_text, best_score = r["doc_id"], r["text"], score

print(f"Best match: {best_text} (Source: {best_id})")
```

### Q13. Function calling (tools)
**Unit 4** | **Marks:** 5 | **Level:** L3 - Apply  
**Dataset file(s):** `q13_sales_data.csv`

Sales data: [100, 250, 175, 300].  
Write three functions: total_sales(data), average_sales(data) and max_sales(data).  
Then write run_tool(command, data) that calls the correct function for 'total', 'average' or 'max' and prints 'Unknown tool' for any other command.  
Test it with 'total', 'max' and 'median'.

```python
import csv

def total_sales(data):
    return sum(data)

def average_sales(data):
    return sum(data) / len(data)

def max_sales(data):
    return max(data)

def run_tool(command, data):
    tools = {"total": total_sales, "average": average_sales, "max": max_sales}
    if command in tools:
        return tools[command](data)
    return "Unknown tool"

with open("q13_sales_data.csv") as f:
    data = [int(r["sales"]) for r in csv.DictReader(f)]

for command in ["total", "max", "median"]:
    print(command, "->", run_tool(command, data))
```

### Q14. Multi-agent workflow
**Unit 4** | **Marks:** 5 | **Level:** L4 - Analyze  
**Dataset file(s):** `q14_raw_data.csv`

Write three functions that act as agents:  
- data_agent(data): removes None values from the list.  
- insight_agent(data): returns the average and the maximum.  
- validation_agent(avg, data): returns 'Valid' if the average lies between the minimum and maximum of the data, otherwise 'Invalid'.  
Run them in sequence on data = [120, None, 150, 130, None, 180] and print the output of each agent.

```python
import csv

def data_agent(values):
    return [float(v) for v in values if v not in (None, "")]

def insight_agent(data):
    return sum(data) / len(data), max(data)

def validation_agent(avg, data):
    return "Valid" if min(data) <= avg <= max(data) else "Invalid"

with open("q14_raw_data.csv") as f:
    raw = [r["value"] for r in csv.DictReader(f)]

clean = data_agent(raw)
print("Data agent output:", clean)
avg, mx = insight_agent(clean)
print("Insight agent output: average =", round(avg, 2), ", maximum =", mx)
print("Validation agent output:", validation_agent(avg, clean))
```

### Q16. Decision matrix
**Unit 5** | **Marks:** 5 | **Level:** L3 - Apply  
**Dataset file(s):** `q16_options.csv, q16_weights.csv`

Three options are scored (1-10) on cost, quality and time: Option A = (7, 8, 6), Option B = (9, 6, 7), Option C = (6, 9, 8).  
Weights: cost = 0.5, quality = 0.3, time = 0.2.  
Write a program to calculate the weighted score of each option and print the best option.

```python
import csv

with open("q16_weights.csv") as f:
    weights = {r["criterion"]: float(r["weight"]) for r in csv.DictReader(f)}

best, best_score = None, -1
with open("q16_options.csv") as f:
    for r in csv.DictReader(f):
        score = sum(float(r[c]) * weights[c] for c in weights)
        print(r["option"], "weighted score:", round(score, 2))
        if score > best_score:
            best, best_score = r["option"], score

print("Best option:", best)
```

### Q17. What-if analysis
**Unit 5** | **Marks:** 5 | **Level:** L3 - Apply  
**Dataset file(s):** `q17_scenarios.csv`

Cost per unit = 60. Scenarios:  
(i) price 90, expected units 600  
(ii) price 100, expected units 500  
(iii) price 110, expected units 430  
Write a program to calculate profit = (price - cost) x units for each scenario and print the best scenario.

```python
import csv

COST = 60
best, best_profit = None, None
with open("q17_scenarios.csv") as f:
    for r in csv.DictReader(f):
        profit = (int(r["price"]) - COST) * int(r["units"])
        print(r["scenario"], "profit:", profit)
        if best_profit is None or profit > best_profit:
            best, best_profit = r["scenario"], profit

print("Best scenario:", best, "with profit", best_profit)
```

### Q18. Responsible AI checklist
**Unit 5** | **Marks:** 5 | **Level:** L4 - Analyze  
**Dataset file(s):** `q18_recommendation.json`

An AI system gives this recommendation: {'decision': 'Reject loan', 'confidence': 0.58, 'explanation': '', 'human_reviewed': False}.  
Write a function that checks three rules: (1) the explanation is not empty, (2) confidence >= 0.70, (3) human_reviewed is True.  
Print PASS or FAIL for each rule. Print 'Recommendation can be used' only if all three pass, otherwise print 'Do not use without review'.

```python
import json

with open("q18_recommendation.json") as f:
    rec = json.load(f)

checks = {
    "Explanation is not empty": rec["explanation"].strip() != "",
    "Confidence >= 0.70": rec["confidence"] >= 0.70,
    "Human reviewed": rec["human_reviewed"] is True,
}
for name, ok in checks.items():
    print(name, ":", "PASS" if ok else "FAIL")

print("Recommendation can be used" if all(checks.values()) else "Do not use without review")
```
